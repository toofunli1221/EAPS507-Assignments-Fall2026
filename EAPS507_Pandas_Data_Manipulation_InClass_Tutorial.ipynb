````python
streams = df[is_stream]
streams[["FEATURE_NAME", "COUNTY_NAME", "ELEV_IN_M"]].head()
````

### `groupby()`
Conceptually, `groupby()` performs **split → apply → combine**:

1. split rows into groups;
2. apply a calculation to each group;
3. combine the group results.

Here we ask: *What is the mean elevation for each county?*

````python
county_elevation = (
    indiana.groupby("COUNTY_NAME")["ELEV_IN_M"]
           .mean()
)

county_elevation.head()
````

### Predict first
What does each row in `county_elevation` represent?

Now predict what changing `.mean()` to `.max()` would mean scientifically.

Each row represents one Indiana county and the mean of its available feature elevations. Replacing `.mean()` with `.max()` gives the highest recorded feature elevation in that county, not necessarily its highest terrain point. Both calculations skip missing elevations and describe the sampled named features.

````python
indiana.groupby("COUNTY_NAME")["ELEV_IN_M"].max().head()
````

### `agg()` — several summaries at once
`agg()` lets us apply multiple summary functions to each group.

````python
feature_elevation = (
    indiana.groupby("FEATURE_CLASS")["ELEV_IN_M"]
           .agg(["count", "mean", "median", "min", "max"])
)

feature_elevation.head(10)
````

- count

- maximum

Hint: the aggregation name for standard deviation is `"std"`.

````python
# Split Indiana records by county and summarize the non-missing elevations.
# count is the number of available elevations; std is the sample standard deviation.
county_stats = indiana.groupby("COUNTY_NAME")["ELEV_IN_M"].agg(
    ["count", "mean", "std", "min", "max"]
)
county_stats.head()
````

### Student-led test 5C — make a scientific comparison
Compare elevation statistics for only `Stream`, `Lake`, and `Summit` features.

### FINAL CHALLENGE — Question 6: Your own question

**My question:**  Which five Indiana counties have the highest mean recorded stream
elevation among counties with at least 10 non-missing stream elevations?

**Why this question is useful/interesting:**  It compares the elevation of recorded stream locations across counties. Requiring at least 10 measurements keeps rankings from being determined by counties with only one or two observations.

**Pandas operations I expect to need:**  Filter Indiana streams > group by county > aggregate record counts and elevation statistics > filter on measurement count > sort by mean elevation > select the top five. This exceeds the three-operation requirement.

````python
# Write your analysis for Question 6 here.
# Restrict to Indiana streams; retain all records for the record-count column.
indiana_streams = indiana.loc[indiana["FEATURE_CLASS"] == "Stream"]
stream_county_stats = indiana_streams.groupby("COUNTY_NAME").agg(
    stream_records=("FEATURE_ID", "size"),
    elevations_available=("ELEV_IN_M", "count"),
    mean_elevation_m=("ELEV_IN_M", "mean"),
    min_elevation_m=("ELEV_IN_M", "min"),
    max_elevation_m=("ELEV_IN_M", "max"),
)

# Apply the minimum number of measurements, then rank and select the five highest means.
top_stream_counties = (
    stream_county_stats.loc[stream_county_stats["elevations_available"] >= 10]
    .sort_values("mean_elevation_m", ascending=False)
    .head(5)
)
top_stream_counties
````
